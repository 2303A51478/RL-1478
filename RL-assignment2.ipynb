```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B - Initialize N, R and Q
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0, 0, 0]

total_reward = 0

# Part C - Epsilon Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    if r < epsilon:
        # Exploration
        selected = random.randint(0, 2)
        print("Exploration")

    else:
        # Exploitation
        selected = Q.index(max(Q))
        print("Exploitation")

    print("Play", play)
    print("Random number:", r)
    print("Selected machine:", machines[selected])

    # Part D - Reward Collection
    reward = float(input("Enter reward: "))

    # Part E - Update N
    N[selected] += 1

    # Update total reward of selected machine
    R[selected] += reward

    # Update Q using incremental sample-average formula
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    # Total reward received
    total_reward += reward

    print("Updated Q-values:", Q)
    print()

# Part F - Display Results

print("================================")
print("FINAL RESULTS")
print("================================")

print("Machine\tN\tR\tQ")

for i in range(3):
    print(machines[i], "\t", N[i], "\t", R[i], "\t", Q[i])

print()
print("Total Reward after 10 plays:", total_reward)
```
