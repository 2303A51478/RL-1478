```python
n = int(input("Enter number of machines: "))

machines = []
q_values = []

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split(',')))

    machines.append(name)

    q = sum(rewards) / len(rewards)
    q_values.append(q)

    print("Machine", name)
    print("Rewards:", rewards)
    print("Q(" + name + ") =", q)

max_q = q_values[0]
selected = machines[0]

for i in range(1, n):
    if q_values[i] > max_q:
        max_q = q_values[i]
        selected = machines[i]

print("Greedy Selected Machine:", selected)
```

```python
n = int(input("Enter number of machines: "))

machines = []
q = [0] * n
count = [0] * n

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split(',')))

    machines.append(name)

    for r in rewards:
        count[i] += 1

        q[i] = q[i] + (r - q[i]) / count[i]

        print("Machine:", name)
        print("Reward:", r)
        print("N =", count[i])
        print("Q =", q[i])
        print()

best = 0

for i in range(1, n):
    if q[i] > q[best]:
        best = i

print("Final Q values:")

for i in range(n):
    print("Q(" + machines[i] + ") =", q[i])

print("Greedy Selected Machine:", machines[best])
```
